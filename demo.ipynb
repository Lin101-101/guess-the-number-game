```python
import random

# 1. 电脑生成一个随机数字
number = random.randint(1, 100)
print("电脑已经生成了一个1到100之间的数字")
```

```python
# 2. 模拟猜数字的过程（为了能在网页上直接展示结果，我们用预设的数字来演示）
guesses = [50, 25, 37, 35]
target_number = 35

for guess in guesses:
    if guess < target_number:
        print(f"你猜了 {guess}，太小了！")
    elif guess > target_number:
        print(f"你猜了 {guess}，太大了！")
    else:
        print(f"你猜了 {guess}，恭喜你猜对了！")
        break
```
